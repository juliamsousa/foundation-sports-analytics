Essa seção consiste no desenvolvimento de um projeto prático de Machine Learning focado na área imobiliária. A seguir estão descritas algumas características do problema:

- Trata-se de um problema **supervisionado**, pois exemplos com um rótulo são dados. Nesse caso os exemplos (instâncias) consistem em distritos com o rótulo contendo o seu preço médio.
- É um problema típico de **regressão**, pois é preciso prever um valor.
- O tipo de regressão é a **univariada**, pois apenas um valor está sendo predito.
- O aprendizado será feito em _batch_ (lotes), uma vez que não há fluxo contínuo de dados entrando no sistema e que não há necessidade de ajustar os dados rapidamente. Além disso, os dados são grandes o suficiente para caber em memória.

Abaixo foi criado um pequeno script que importa os dados para o problema. Ele realiza o download dos dados zipados a partir de uma url e realiza a descompressão dos dados na pasta destino.

```python
# Importação das bibliotecas necessárias
import os
import tarfile
from six.moves import urllib
```

```python
# Definição de variáveis constantes
DOWNLOAD_ROOT = "https://raw.githubusercontent.com/ageron/handson-ml2/master/"
HOUSING_PATH = os.path.join("datasets", "housing")
HOUSING_URL = DOWNLOAD_ROOT + "datasets/housing/housing.tgz"
```

```python
# Definição da função que faz o download dos dados e descompacta na pasta destino
def fetch_housing_data(housing_url=HOUSING_URL, housing_path=HOUSING_PATH):
    if not os.path.isdir(housing_path):
        os.makedirs(housing_path)
        tgz_path = os.path.join(housing_path, "housing.tgz")
        urllib.request.urlretrieve(housing_url, tgz_path)
        housing_tgz = tarfile.open(tgz_path)
        housing_tgz.extractall(path=housing_path)
        housing_tgz.close()
```

```python
# Chamada da função para construção do dataset
fetch_housing_data()
```
